### Sadly pg_parquet extention is too hard to install to vanila PG, so im doing some parquet -> csv -> PG process.

#### After that .csv is 'hdfs dfs -get'

```python
from pyspark import SparkConf
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("parquetToCSVToPG").master("local[8]").getOrCreate()
sc = spark.sparkContext
```

```python
df = spark.read.parquet("hdfs://hdfs-namenode:9000/partners")
```

```python
df.coalesce(1).write.mode("overwrite").options(header=True, sep="\t").csv("hdfs://hdfs-namenode:9000/To_PG/")
```
